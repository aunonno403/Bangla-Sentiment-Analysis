# Preprocessing Test

Test the preprocessing pipeline on `data/processed/train.csv`. Verify:
- NaN handling
- Text normalization and cleaning
- Tokenization and stemming
- TF-IDF feature extraction

```python
# Cell 1: Imports and setup
import sys
sys.path.insert(0, '../src')

import pandas as pd
import logging
from data_preprocessing import preprocess_text, preprocess_dataframe, create_tfidf_features

logging.basicConfig(level=logging.WARNING)
print("Imports successful - now using bnlp for tokenization")
```

```python
# Cell 2: Load train.csv
train = pd.read_csv('../data/processed/train.csv')
print(f"Loaded train.csv: {train.shape}")
print(f"Columns: {train.columns.tolist()}")
print(f"\nFirst few rows:")
train.head()
```

```python
# Cell 3: Test preprocessing on sample texts
print("Testing preprocessing on sample texts (bnlp tokenizer):\n")
for i, row in train.head(3).iterrows():
    text = row['Comment']
    print(f"Original: {text[:80]}...")
    cleaned = preprocess_text(text)
    print(f"Cleaned:  {cleaned[:80]}...")
    print()
```

```python
# Cell 4: Apply preprocessing to sample (500 rows)
train_sample = train.head(500).copy()
print(f"Processing {len(train_sample)} rows...")

train_processed = preprocess_dataframe(train_sample, text_column='Comment')
print(f"Preprocessed shape: {train_processed.shape}")
print(f"Cleaned text sample (first 3):")
for idx, text in enumerate(train_processed['cleaned_text'].head(3), 1):
    print(f"  {idx}. {text[:100]}")
```

```python
# Cell 5: Test TF-IDF feature extraction
vectorizer, X_train, _ = create_tfidf_features(train_processed, max_features=1000)
print(f"TF-IDF vectorizer created")
print(f"  Features: {len(vectorizer.get_feature_names_out())}")
print(f"  X_train shape: {X_train.shape}")
print(f"  Sparsity: {(1 - X_train.nnz / (X_train.shape[0] * X_train.shape[1])) * 100:.2f}%")  # type: ignore
```

```python
# Cell 6: Process full train and test sets and save
print("Processing full train set (this may take 1-2 minutes)...")
train_full = preprocess_dataframe(train, text_column='Comment')
train_full.to_csv('../data/processed/train_cleaned.csv', index=False)
print(f"Saved train: {len(train_full)} rows to data/processed/train_cleaned.csv")

# Also process test set
print("Processing test set...")
test = pd.read_csv('../data/processed/test.csv')
test_full = preprocess_dataframe(test, text_column='Comment')
test_full.to_csv('../data/processed/test_cleaned.csv', index=False)
print(f"Saved test: {len(test_full)} rows to data/processed/test_cleaned.csv")

print("\n✓ Preprocessing complete!")
```
